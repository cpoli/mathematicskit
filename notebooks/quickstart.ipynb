```python
import matplotlib.pyplot as plt
import numpy as np

import mathematicskit as mk

print(mk.__version__)
```

## 1. Root finding: watch Newton's method converge quadratically

`solve()` returns a `RootResult` whose `history` holds every iterate, so the convergence rate is something you can *see*.

```python
from mathematicskit.numerical_analysis.visualizers import plot_convergence_history

f, fprime = (lambda x: x**2 - 2.0), (lambda x: 2.0 * x)
newton = mk.numerical_analysis.NewtonRaphson(f, fprime, x0=3.0, tol=1e-14).solve()
bisection = mk.numerical_analysis.Bisection(f, 0.0, 3.0, tol=1e-14).solve()
print(f"Newton: {len(newton.history) - 1} steps, bisection: {len(bisection.history) - 1} steps")

ax = plot_convergence_history(bisection, root_exact=np.sqrt(2))
plot_convergence_history(newton, root_exact=np.sqrt(2), ax=ax)
ax.legend();
```

## 2. Fractals: the Mandelbrot set

The escape-time kernel is JIT-compiled by numba when it is installed, and runs as plain NumPy/Python otherwise.

```python
from mathematicskit.fractals_chaos.visualizers import plot_escape_time

plot_escape_time(mk.fractals_chaos.mandelbrot_set(resolution=400, max_iter=100), cmap="magma");
```

## 3. Dynamical systems: the logistic map's route to chaos

```python
from mathematicskit.ode_dynamics.visualizers import plot_bifurcation_diagram

r_plot, x_plot = mk.ode_dynamics.bifurcation_diagram(np.linspace(2.8, 4.0, 800), n_keep=100)
plot_bifurcation_diagram(r_plot, x_plot);
```

## 4. ODEs: the Lorenz attractor

```python
lorenz = mk.ode_dynamics.LorenzSystem([1.0, 1.0, 1.0]).integrate((0.0, 40.0), dt=1e-2, method="rk4")
x, y, z = lorenz.y.T
plt.figure().add_subplot(projection="3d").plot(x, y, z, lw=0.5);
```

## 5. Number theory: exact integer arithmetic

```python
print(mk.number_theory.is_prime_miller_rabin(2**61 - 1))  # a Mersenne prime
print(mk.number_theory.prime_factorization(2**32 + 1))  # Euler's factorization of the 5th Fermat number
```

## 6. Graph theory: shortest paths without networkx

```python
graph = mk.graph_theory.Graph(4)
for u, v, w in [(0, 1, 1.0), (1, 2, 2.0), (0, 2, 5.0), (2, 3, 1.0)]:
    graph.add_edge(u, v, w)
result = mk.graph_theory.dijkstra_shortest_paths(graph, sources=0)
print(result.distances)  # 0 -> 2 goes via 1 (cost 3), not directly (cost 5)
```
